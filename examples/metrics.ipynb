# Example: Metric configs and how to define new ones

The following notebook provides tutorial and code for defining metrics.

```python
from garisom_tools.utils.metric import Metric, get_mapping_dict
from garisom_tools.config.metric import MetricConfig

import numpy as np
```

## Defining new metrics

```python
# Write metric function
def mean_absolute_error(targets, prediction) -> float:
    """Defines Mean Absolute Error"""
    return np.sum(np.abs(prediction - targets)) / len(targets)
```

```python
# Create new metric class
class MAE(Metric):
    def __init__(self, output_name: str, name: str = "mae"):
        super().__init__(name=name, output_name=output_name, func=mean_absolute_error)
```

```python
# Add to the mapping dict
mappings = get_mapping_dict()
mappings['mae'] = MAE
```

```python
# Create a metric config with new metrics
metric_config_dict = {
    'params': ['leaftemp'], 
    'metrics': ['mae'],
    'modes': ['min']
}
metric_config = MetricConfig.from_dict(metric_config_dict)
```
